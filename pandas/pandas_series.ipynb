```python
import numpy as np
import pandas as pd
```

```python
#series from list 
#string
country = ['India', 'usa', 'Nepal', 'Russia']
pd.Series(country)
```

```python
#custom index
marks = [67,57, 89,100]
subjects = ['math', 'english', 'science' , 'hindi']
pd.Series(marks, index= subjects)
```

```python
marks = pd.Series(marks, index= subjects, name = "ayush marks")
marks
```

```python
#series from dict
marks = {
    'math': 76,
    'eng': 87,
    'science': 89,
    'hindi': 99
} 
marks_series= pd.Series(marks , name = "ayush")
marks_series
```
